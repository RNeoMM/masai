```python
#profiling

#%timeit --> Quick Performance Checks

arr = np.arange(1000000)

#python loops
def squares(a):
  result = []
  for x in a:
    result.append(x ** 2)
  return result

#numpy
def square_numpy(a):
  return a ** 2
```

```python
%timeit squares(arr)
%timeit square_numpy(arr)
```

```python
#using Time Module
import time

arr = np.arange(1000000)

strt = time.time()
result = np.square(arr)
end = time.time()

print(f"Execution Time: {end - strt:.5f} seconds")
```

```python
#cprofile --> shows how much time each function in your program takes

import cProfile

def compute():
  x = np.random.rand(1000,1000)
  y = np.dot(x,x.T)
  z = np.linalg.inv(y[:100,:100])
  return z.sum()
```

```python
cProfile.run('compute()')
```
